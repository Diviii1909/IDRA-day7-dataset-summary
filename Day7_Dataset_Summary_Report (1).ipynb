## 11. Check for Missing Values

Missing values are checked column-by-column using `isna().sum()`.

```python
missing_values = df.isna().sum()

print("Missing values in each column:")
display(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

## 12. Basic Dataset Observations

Based on the Pandas exploration above, the following observations can be made:

```python
print(f"• The dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")
print(f"• The available features are: {', '.join(df.columns)}.")
print("• Integer columns:", ", ".join(df.select_dtypes(include="int64").columns))
print("• Floating-point columns:", ", ".join(df.select_dtypes(include="float64").columns))
print("• Text/object columns:", ", ".join(df.select_dtypes(include="object").columns))

missing = df.isna().sum()
nonzero_missing = missing[missing > 0]

if len(nonzero_missing) == 0:
    print("• No missing values were found.")
else:
    print("• Missing values were found in:")
    for column, count in nonzero_missing.items():
        print(f"  - {column}: {count} missing value(s)")

print(f"• The average age is {df['Age'].mean():.2f} years.")
print(f"• The average marks value is {df['Marks'].mean():.2f} (calculated from non-missing values).")
print(f"• The average attendance is {df['Attendance_Percentage'].mean():.2f}% (calculated from non-missing values).")
print(f"• The dataset contains {df['Course'].nunique()} unique courses and {df['City'].nunique()} unique cities.")
```
